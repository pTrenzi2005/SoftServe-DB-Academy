Declarative pipeline (materialized_view) that loads the csv file into the bronze schema.

```python
import pyspark.sql.functions as F
from pyspark import pipelines

@pipelines.materialized_view(
    name="trezio2005_bronze.bronze_customers"
)
def create_bronze_customers():
    return (
        spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load("/Volumes/samples/databricks/datasets/retail-org/customers/customers.csv")
    )
```
