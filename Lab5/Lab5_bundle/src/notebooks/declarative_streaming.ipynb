Declarative pipelines use decorators that register code as declarative data target. This approach allows to skip the manual definition of paths, checkpoints and specific instructions. 

This specific code enables autoloader (option cloudFiles) and sets it up to monitor the specified data folder, everything else is set up under the hood. I used data provided by the databricks that simulates the scenario of selling products to consumers.


```python
import pyspark.sql.functions as F
from pyspark import pipelines

@pipelines.table(
    name="trezio2005_bronze.bronze_sales_orders"
)
def create_bronze_sales():
    return (
        spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "json")
        .option("cloudFiles.inferColumnTypes", "true")
        .load("/Volumes/samples/databricks/datasets/retail-org/sales_stream/sales_stream.json/")
    )
```
