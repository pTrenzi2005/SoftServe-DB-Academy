Creating schemas for the free environment

```python
dbutils.widgets.combobox("catalog", "workspace", ["workspace", "dbr_dev"])
catalog_name = dbutils.widgets.get("catalog")

spark.sql(f"""
          CREATE SCHEMA IF NOT EXISTS {catalog_name}.trezio2005_bronze
          """)
spark.sql(f"""
          CREATE SCHEMA IF NOT EXISTS {catalog_name}.trezio2005_silver
          """)
```

Ingesting data into the bronze layer

```python
volume_path = f"/Volumes/{catalog_name}/trezio2005_bronze/volume/Mens-Volleyball-PlusLiga-2008-2023.csv"
table_name = f"{catalog_name}.trezio2005_bronze.Mens_Volleyball_raw"

df_bronze = (spark.read.format("csv") 
                .option("header", "true") 
                .option("inferSchema", "true") 
                .load(volume_path)
            )

(df_bronze.write.format("delta") 
    .mode("append") 
    .option("overwriteSchema", "true")
    .saveAsTable(table_name)
)
```
